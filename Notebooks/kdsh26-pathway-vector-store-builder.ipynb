```python
# ============================================================================
# NOTEBOOK 1: PATHWAY VECTOR STORE BUILDER & STORAGE
# Using Pathway for ALL operations (no manual file I/O)
# ============================================================================


import pathway as pw
from pathway.xpacks.llm import embedders
import pandas as pd
import numpy as np
from transformers import AutoTokenizer, AutoModel
import torch
import json
from datetime import datetime
import os
```

```python
# ============================================================================
# STEP 1: Configuration
# ============================================================================

BOOKS = {
    'In-search-of-the-castaways': '/kaggle/input/kharagpur-data-science-hackathon-kdsh-2026-dataset/Books/In search of the castaways.txt',
    'The-count-of-monte-cristo': '/kaggle/input/kharagpur-data-science-hackathon-kdsh-2026-dataset/Books/The Count of Monte Cristo.txt'
}

OUTPUT_DIR = '/kaggle/working'
PATHWAY_STORAGE_DIR = '/kaggle/working/pathway_storage'

# Create storage directory
os.makedirs(PATHWAY_STORAGE_DIR, exist_ok=True)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"✅ Using device: {device}")
```

```python
# ============================================================================
# STEP 3: Pathway Document Loader & Transformer
# ============================================================================

class PathwayDocumentProcessor:
    """Process books using Pathway transformations."""
    
    @staticmethod
    def load_books(books_dict):
        """Load books and create Pathway tables."""
        
        print("\n📚 Loading books with Pathway...")
        
        all_chunks = []
        
        for book_name, book_path in books_dict.items():
            print(f"\n📖 Processing: {book_name}")
            
            # Load book
            with open(book_path, 'r', encoding='utf-8', errors='ignore') as f:
                text = f.read()
            
            print(f"   Length: {len(text):,} characters")
            
            # Chunk using Pathway-compatible method
            paragraphs = [p.strip() for p in text.split('\n\n') if p.strip()]
            
            chunks = []
            current = []
            current_len = 0
            chunk_id = 0
            
            for para in paragraphs:
                words = para.split()
                para_len = len(words)
                
                if current_len + para_len <= 500:
                    current.extend(words)
                    current_len += para_len
                else:
                    if current:
                        chunk_text = ' '.join(current)
                        chunks.append({
                            'chunk_id': chunk_id,
                            'book': book_name,
                            'chunk': chunk_text,
                            'chunk_length': len(current),
                            'created_at': datetime.now().isoformat(),
                            'is_valid': len(chunk_text) > 50
                        })
                        chunk_id += 1
                    
                    if len(current) > 100:
                        current = current[-100:] + words
                        current_len = len(current)
                    else:
                        current = words
                        current_len = para_len
            
            if current:
                chunk_text = ' '.join(current)
                chunks.append({
                    'chunk_id': chunk_id,
                    'book': book_name,
                    'chunk': chunk_text,
                    'chunk_length': len(current),
                    'created_at': datetime.now().isoformat(),
                    'is_valid': len(chunk_text) > 50
                })
            
            all_chunks.extend(chunks)
            print(f"   ✅ Created {len(chunks)} chunks")
        
        # Create Pathway table from chunks
        chunks_df = pd.DataFrame(all_chunks)
        chunks_table = pw.debug.table_from_pandas(chunks_df)
        
        print(f"\n✅ Total chunks in Pathway: {len(chunks_df)}")
        
        return chunks_table, chunks_df
```

```python
# ============================================================================
# STEP 2: Pathway Document Schema & Chunker
# ============================================================================

# Define Pathway schema for documents
class DocumentSchema(pw.Schema):
    """Schema for document chunks in Pathway."""
    chunk_id: int
    book: str
    chunk: str
    chunk_length: int
    created_at: str
    is_valid: bool

class ChunkWithEmbeddingSchema(pw.Schema):
    """Schema for chunks with embeddings."""
    chunk_id: int
    book: str
    chunk: str
    chunk_length: int
    embedding: list  # Will store as list in Pathway
    embedding_dim: int
    score: float 
    created_at: str

class RetrievalResultSchema(pw.Schema):
    """Schema for retrieval results."""
    query_id: int
    chunk_id: int
    book: str
    chunk: str
    score: float
    rank: int
```

```python
# ============================================================================
# UPDATED EMBEDDING GENERATOR FOR H100 - Using NVIDIA NV-Embed-v2
# ============================================================================

class PathwayEmbeddingTransformer:
    """Transform documents into embeddings using NVIDIA NV-Embed-v2 (H100 optimized)."""
    
    def __init__(self, model_name="nvidia/NV-Embed-v2"):
        """
        Initialize with NVIDIA's NV-Embed-v2 - optimized for H100
        
        Model options:
        - "nvidia/NV-Embed-v2" ⭐ BEST for H100 (4096 dims, superior quality)
        - "nvidia/NV-Embed-v1" (older version)
        - "BAAI/bge-m3" (if you need multilingual, 1024 dims)
        
        NV-Embed-v2 specs:
        - Embedding dimension: 4096
        - Max sequence length: 32768
        - Batch size: Can handle 256+ on H100
        - Context: Supports long documents
        """
        print(f"\n📦 Loading NVIDIA model: {model_name}")
        print(f"   Dimension: 4096")
        print(f"   GPU: H100 (perfect fit!)")
        
        self.model_name = model_name
        self.embedding_dim = 4096
        self.device = device
        
        try:
            # For NV-Embed-v2, use sentence-transformers for compatibility
            from sentence_transformers import SentenceTransformer
            self.model = SentenceTransformer(model_name, device=device)
            print(f"✅ Model loaded successfully!")
        except Exception as e:
            print(f"⚠️ Error loading {model_name}: {e}")
            print("Fallback: Using bge-m3")
            from sentence_transformers import SentenceTransformer
            self.model = SentenceTransformer("BAAI/bge-m3", device=device)
            self.embedding_dim = 1024
    
    def embed_chunks_table(self, chunks_table, chunks_df, batch_size=128):
        """
        Generate embeddings for chunks using H100.
        
        With H100:
        - Can use batch_size=256+ for speed
        - Processes entire dataset in seconds
        - NV-Embed-v2 gives superior quality
        """
        
        print(f"\n🔄 Generating {self.model_name} embeddings...")
        print(f"   Batch size: {batch_size} (H100 optimized)")
        print(f"   Total chunks: {len(chunks_df)}")
        
        chunk_texts = chunks_df['chunk'].tolist()
        embeddings = []
        
        total_batches = (len(chunk_texts) + batch_size - 1) // batch_size
        
        for i in range(0, len(chunk_texts), batch_size):
            batch = chunk_texts[i:i+batch_size]
            
            # Use sentence-transformers encode (optimized for GPU)
            batch_embeddings = self.model.encode(
                batch,
                batch_size=batch_size,
                show_progress_bar=False,
                convert_to_numpy=True,
                normalize_embeddings=True  # Important for semantic search
            )
            
            embeddings.extend(batch_embeddings.tolist())
            
            batch_num = i // batch_size + 1
            if batch_num % 5 == 0 or batch_num == total_batches:
                print(f"   Batch {batch_num}/{total_batches} ✅")
        
        # Add necessary columns to match your schema
        chunks_df['embedding'] = embeddings
        chunks_df['embedding_dim'] = chunks_df['embedding'].apply(lambda x: len(x))
        chunks_df['score'] = 0.0 
        
        # Ensure only columns in the schema are passed
        schema_cols = ['chunk_id', 'book', 'chunk', 'chunk_length', 'embedding', 'embedding_dim', 'score', 'created_at']
        final_df = chunks_df[schema_cols]
        
        print(f"\n✅ Embeddings generated. Creating Pathway table...")
        
        # FIX: Change pw.from_pandas to pw.debug.table_from_pandas
        embeddings_table = pw.debug.table_from_pandas(final_df, schema=ChunkWithEmbeddingSchema)
        
        return embeddings_table, final_df
```

```python
# ============================================================================
# STEP 5: Pathway Vector Store Manager
# ============================================================================

class PathwayVectorStoreManager:
    """Manage vector stores using Pathway tables and persistent storage."""
    
    def __init__(self, storage_dir):
        self.storage_dir = storage_dir
        self.tables = {}
        self.metadata = {
            'created_at': datetime.now().isoformat(),
            'model': 'sentence-transformers/all-MiniLM-L6-v2',
            'embedding_dim': 384,
            'cost': '$0 (FREE)',
            'books': {},
            'pathway_tables': {}
        }
    
    def save_vector_store(self, book_name, embeddings_table, chunks_df):
        """Save vector store using Pathway's persistence."""
        
        print(f"\n💾 Saving vector store for {book_name}...")
        
        # Store Pathway table
        self.tables[book_name] = embeddings_table
        
        # Save embeddings as numpy
        embeddings_array = np.array([e for e in chunks_df['embedding'].values])
        emb_path = f"{self.storage_dir}/embeddings_{book_name}.npy"
        np.save(emb_path, embeddings_array)
        
        # Save chunks CSV
        chunks_csv = f"{self.storage_dir}/chunks_{book_name}.csv"
        chunks_df_clean = chunks_df.drop('embedding', axis=1)
        chunks_df_clean.to_csv(chunks_csv, index=False)
        
        # Save Pathway table as parquet (Pathway native format)
        parquet_path = f"{self.storage_dir}/pathway_chunks_{book_name}.parquet"
        # embeddings_table.to_pandas().to_parquet(parquet_path)
        chunks_df.to_parquet(parquet_path, index=False)
        
        # Update metadata
        self.metadata['books'][book_name] = {
            'embeddings_npy': emb_path,
            'chunks_csv': chunks_csv,
            'num_chunks': len(chunks_df),
            'embedding_dim': 384
        }
        
        self.metadata['pathway_tables'][book_name] = {
            'parquet_path': parquet_path,
            'table_type': 'ChunkWithEmbeddingSchema'
        }
        
        print(f"✅ Saved embeddings: {emb_path}")
        print(f"✅ Saved chunks CSV: {chunks_csv}")
        print(f"✅ Saved Pathway table: {parquet_path}")
    
    def save_metadata(self):
        """Save metadata as JSON."""
        
        metadata_path = f"{self.storage_dir}/vector_store_metadata.json"
        
        with open(metadata_path, 'w') as f:
            json.dump(self.metadata, f, indent=2)
        
        print(f"\n✅ Metadata saved: {metadata_path}")
        
        return metadata_path
```

```python
# ============================================================================
# UPDATED MAIN EXECUTION (FIXED)
# ============================================================================

print("\n" + "=" * 100)
print("🚀 PATHWAY VECTOR STORE BUILDER - H100 OPTIMIZED")
print("=" * 100)

# Step 1: Load books and create chunks table
processor = PathwayDocumentProcessor()
chunks_table, chunks_df = processor.load_books(BOOKS)

print("\n📊 Chunks Table Schema:")
print(chunks_table.schema)

# Step 2: Create embeddings using NV-Embed-v2 on H100
embedder = PathwayEmbeddingTransformer(model_name="nvidia/NV-Embed-v2")
# Ensure your embed_chunks_table function inside Step 4 uses pw.debug.table_from_pandas as well
embeddings_table, enriched_df = embedder.embed_chunks_table(chunks_table, chunks_df, batch_size=256)

print("\n📊 Embeddings Table Schema:")
print(embeddings_table.schema)

# Step 3: Save to persistent storage using Pathway
manager = PathwayVectorStoreManager(PATHWAY_STORAGE_DIR)

for book_name in BOOKS.keys():
    # Filter for the specific book and reset index
    book_chunks = enriched_df[enriched_df['book'] == book_name].reset_index(drop=True)
    
    # FIX: Use pw.debug.table_from_pandas with the correct Schema
    book_table = pw.debug.table_from_pandas(book_chunks, schema=ChunkWithEmbeddingSchema)
    
    manager.save_vector_store(book_name, book_table, book_chunks)
```
